```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
#import data frames
```

```python
X_train = pd.read_csv('X_train_scaled.csv')
X_val = pd.read_csv('X_val_scaled.csv')

y_train = pd.read_csv('y_train.csv')
y_train = y_train.values.ravel() #turns data frame into 1d from 2d
y_val = pd.read_csv('y_val.csv')
y_val = y_val.values.ravel()#turns data frame into 1d from 2d
```

Build Baseline model with default parameters.

```python
dtc = DecisionTreeClassifier()
```

```python
dtc.fit(X_val, y_val)
```

```python
y_pred = dtc.predict(X_val)
```

```python
print(confusion_matrix(y_val, y_pred)) #first row: true +ve ad false+ve. second row: false -ve and true -ve
```

```python
print(classification_report(y_val, y_pred))
```

Default model has no false -ve/+ve. Precision and Recall is 100 percent. I think the default model is the best.

```python
dtc.score(X_train,y_train)
```

```python
dtc.score(X_val,y_val)
```

```python
dtc.feature_importances_ #shows an array of how important each feature is
```

```python
features = pd.DataFrame(dtc.feature_importances_, index = X_train.columns)
```

```python
features.head(15)
```

inspection_season_summer was the msot important feature.
