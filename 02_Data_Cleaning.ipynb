Data cleaning to prepair data.

```python
%pip install pandas numpy matplotlib seaborn
import pandas as pd
import numpy as np 
```

```python
#Load data set
df = pd.read_csv('urbandata.csv')
```

```python
df.info()
```

Describe every column and check for inconsistencies and flag them

Dropping any rows with missing values

```python
df.isna().any(axis=1).sum()
```

```python
df[df.isna().any(axis=1)]
```

```python
df = df.dropna()
```

```python
df.shape
```

Removing misspellings

```python
df[["inspection_season","inspection_window"]].value_counts()
```

```python
df = df.replace(to_replace ="sumer", value ="summer")
```

```python
df = df.replace(to_replace="morni7ng", value="morning")
```

```python
df[["inspection_season","inspection_window"]].value_counts()
```

Check for any duplicates

```python
df.duplicated().sum()
```

Dropping inspector_id 

```python
df = df.drop("inspector_id", axis=1)
```

```python
df.shape
```

```python
#save cleaned csv before we start splitting and encoding 
```

```python
df.to_csv("urbandata_cleaned.csv", index=False)
```
