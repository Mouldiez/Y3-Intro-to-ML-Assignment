```python
import numpy as np
import pandas as pd

%pip install scikit-learn
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(handle_unknown = 'ignore' , sparse_output = False).set_output(transform = 'pandas')
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
#load in cleaned data
```

```python
df = pd.read_csv('urbandata_cleaned.csv')
```

```python
df.shape
```

```python
df.info
```

Spliting dataframe

```python
#Spliting features (X) from target(Y)
X = df.drop(columns =['failure'])
y = df['failure']            
```

```python
X.shape
```

Train and Test 


```python
# This will be first split 70% to 30% and then we split test again for validation and test.
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, random_state = 29, stratify = y, test_size = 0.3) # the random state has no significance apart from keeping the randomness of the split the same. Stratify just makes sure the failures to nonf ailiures are proportional to thta of the original data.
```

```python
X_train.shape;X_train.describe()
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
X_test, X_val, y_test, y_val = train_test_split(X_test,y_test, random_state = 29, stratify=y_test, test_size = 0.5)
```

```python
#now data is split 70 training, 15 validating, 15 testing.
```

```python
y_val.shape
```

Encoding categorical values

```python
df['inspection_season'].unique()
```

```python
ohetransform = ohe.fit_transform(X_train[['inspection_season']])
```

```python
ohetransform
```

season concat encoded columns and remove original non encded columns

```python
X_train = pd.concat([X_train.reset_index(drop=True), ohetransform.reset_index(drop=True)], axis=1).drop(columns='inspection_season')
```

```python
ohetransform = ohe.transform(X_val[['inspection_season']])
```

```python
X_val = pd.concat([X_val.reset_index(drop=True), ohetransform.reset_index(drop=True)], axis=1).drop(columns='inspection_season')
```

```python
ohetransform = ohe.transform(X_test[['inspection_season']])
```

```python
X_test = pd.concat([X_test.reset_index(drop=True), ohetransform.reset_index(drop=True)], axis=1).drop(columns='inspection_season')
```

```python
df['inspection_window'].unique()
```

```python
ohetransform = ohe.fit_transform(X_train[['inspection_window']])
```

```python
ohetransform
```

window concat encoded rows and remove original non encoded rows

```python
X_train = pd.concat([X_train.reset_index(drop=True), ohetransform.reset_index(drop=True)], axis=1).drop(columns='inspection_window')
```

```python
ohetransform = ohe.transform(X_val[['inspection_window']])
```

```python
X_val = pd.concat([X_val.reset_index(drop=True), ohetransform.reset_index(drop=True)], axis=1).drop(columns='inspection_window')
```

```python
ohetransform = ohe.transform(X_test[['inspection_window']])
```

```python
X_test = pd.concat([X_test.reset_index(drop=True), ohetransform.reset_index(drop=True)], axis=1).drop(columns='inspection_window')
```

Standardising the values

```python
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled =pd.DataFrame(X_train_scaled, columns = ['material_flexibility','structural_mass_ratio','recent_precipitation','infrastructure_age','surface_debris',
'vibration_exposure','asset_density','moisture_stress','inspection_season_autumn','inspection_season_spring','inspection_season_summer','inspection_season_winter',
'inspection_window_afternoon','inspection_window_morning','inspection_window_night'])

X_val_scaled =pd.DataFrame(X_val_scaled, columns = ['material_flexibility','structural_mass_ratio','recent_precipitation','infrastructure_age','surface_debris',
'vibration_exposure','asset_density','moisture_stress','inspection_season_autumn','inspection_season_spring','inspection_season_summer','inspection_season_winter',
'inspection_window_afternoon','inspection_window_morning','inspection_window_night'])

X_test_scaled = pd.DataFrame(X_test_scaled, columns = ['material_flexibility','structural_mass_ratio','recent_precipitation','infrastructure_age','surface_debris',
'vibration_exposure','asset_density','moisture_stress','inspection_season_autumn','inspection_season_spring','inspection_season_summer','inspection_season_winter',
'inspection_window_afternoon','inspection_window_morning','inspection_window_night'])
```
